# **DATA FRAME: Đọc - ghi file**

**1. Đọc ghi file trực tiếp**

```python
import pandas as pd
```

```python
# Khởi tạo dữ liệu dạng dictionary
data = {
    'name': ['Minh Anh', 'Lâm Khánh', 'Tuấn Linh', 'Duy Anh', 'Thu Hương'],
    'age': [15, 15, 16, 17, 17],
    'gender': ['male', 'male', 'male', 'male', 'female']
}
# Khởi tạo DataFrame
df = pd.DataFrame(data)
# Hiển thị dữ liệu từ dataframe
df
```

**Tạo và đọc dữ liệu**

```python
# Tạo dữ liệu CSV
df.to_csv('data.csv', index=False)

# Đọc dữ liệu
df_csv = pd.read_csv('data.csv')
df_csv
```

```python
# Tải file về máy
from google.colab import files
files.download('data.csv')
```

```python
# Tạo dữ liệu xlsx
df = pd.DataFrame(data)
df.to_excel('output.xlsx', index=False)
# Đọc dữ liệu từ file xlsx
df_read = pd.read_excel('output.xlsx', sheet_name='Sheet1')
print(df_read)
```

```python
# Tạo dữ liệu json
df = pd.DataFrame(data)
df.to_json('output.json', orient='records')
# Đọc dữ liệu từ file json
df_read = pd.read_json('output.json')
print(df_read)
```

# **TRUY CẬP DỮ LIỆU**

**1. Truy cập theo cột: DataFrame[column]**

```python
# Truy cập cột 'animal' của df
animals = df['animal']
animals.head(3)
```

```python
# Hiển thị 3 bản ghi cuối của cột 'sound'
sounds = df['sound']
sounds.tail(3)
```

Lưu ý: NaN là viết tắt của Not a Number. <br>
Trong pandas, NaN dùng để biểu diễn dữ liệu bị thiếu (missing data)



```python
# Truy cập nhiều cột: DataFrame[[col1, col2, ...]]
multi_col = df[['animal', 'sound']].tail(3)
multi_col
```

**2. Truy cập theo hàng**

slice[start:end] => chạy từ start đến end-1

```python
# Hiện từ hàng 5 đến 8
animals[4:8]
```

```python
# Hiện 4 hàng đầu tiên
animals[:4]
```

```python
# Hiện từ dòng số 10 đến hết
animals[9:]
```

**3. Truy cập theo cả hàng và cột**

DataFrame.iloc[rows, columns]

```python
# Hiển thị 3 dòng đầu tiên của cột thứ 2
df.iloc[:3, 1:2]
```

**4. Truy cập có điều kiện**

```python
# Tìm bản ghi mà data cột 'animal' == 'cat'
target = df[df['animal'] == 'cat']
target
```

# **Tổng hợp 1 số lệnh hay dùng với DataFrame**

```python
# Khởi tạo dữ liệu dạng dictionary
data = {
    'name': ['Minh Anh', 'Lâm Khánh', 'Tuấn Linh', 'Duy Anh', 'Thu Hương'],
    'age': [15, 15, 16, 17, 17],
    'gender': ['male', 'male', 'male', 'male', 'female']
}
# Khởi tạo DataFrame
df = pd.DataFrame(data)
# Hiển thị dữ liệu từ dataframe
df
```

```python
# Xem kiểu dữ liệu của từng cột
df.dtypes
```

```python
# Xem số hàng và cột
df.shape
```

```python
# Tóm tắt thông tin của DataFrame: số dòng, số cột, số giá trị có dữ liệu, kiểu dữ liệu
df.info()
```

```python
# Thống kê các cột dạng số
    # count: số dữ liệu (bản ghi)
    # mean: trung bình
    # std: độ lệch chuẩn
    # min, max: giá trị lớn/nhỏ nhất
df.describe()
```

```python
# Hiển thị các dòng đầu: df.head(n)
df.head(2)
```

```python
# Hiển thị các dòng cuối: df.tail(n)
df.tail(3)
```

```python
# Đổi tên trục: rename_axis()
    # index (axis=0) → trục hàng
    # columns (axis=1) → trục cột
df = df.rename_axis(index=None) # Xóa tên cột index
df
```

```python
# Đặt 1 cột làm index: set_index()
df = df.set_index('name')
df
```

```python
# Đưa index trở lại thành cột: reset_index()
df.reset_index()
```

```python
# Đổi tên cột hoặc index: df.rename(columns={"old_name":"new_name"})
df = df.rename(columns={"age":"tuổi"})
df
```
