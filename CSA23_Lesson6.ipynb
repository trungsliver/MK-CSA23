# **Thư viện Matplotlib**

- Khái niệm: thư viện phổ biến để tạo các loại biểu đồ
- 1 số thuộc thính hay dùng:
<br>
      + figure(figsize=(h,w)): tạo biểu đồ với kích thước tùy chỉnh, đơn vị inch
      <br>
      + legend: nhãn/label của các cột
      <br>
      + xlabel, ylabel: nhãn của trục x và trục y
      <br>
      + title: tên biểu đồ
      <br>
- Sau khi chỉnh sửa, dùng plt.show() để hiển thị biểu đồ

```python
# Thêm thư viện để chỉnh sửa thuộc tính của biểu đồ
import matplotlib.pyplot as plt
```

```python
# figure(figsize=(h,w)): tạo biểu đồ và thay đổi kích thước (đợn vị inch)
plt.figure(figsize=(14,6))
df_line = df.set_index('Date')
plt.plot(df_line)

# legend: hiển thị nhã của các cột
# xlabel, ylabel hiển thị nhãn của các cột
plt.legend(df_line.columns)
plt.ylabel('Views')

# title: hiển thị tên biểu đồ
plt.title('Spotify song views from January 2017 to January 2018')

# HIển thị biểu đồ
plt.show()
```

# **Biểu đồ cột - Bar Chart**
- Biểu diễn thông tin riêng lẻ rõ ràng
- Tập trung vào các giá trị riêng lẻ

```python
files.upload()    # Billionaires.csv
```

```python
df = pd.read_csv('Billionaires.csv')
df
```

```python
# Làm sạch dữ liệu & chuyển đổi kiểu dữ liệu của cột 'Net worth'
def clean_networth(networth):
    return networth[1:-1]

# df["Net Worth"] = df["Net Worth"].apply(lambda x: x[1:-1])
df["Net Worth"] = df["Net Worth"].apply(clean_networth)
df["Net Worth"] = pd.to_numeric(df["Net Worth"])
df.head()
```

```python
# Tính tổng Net Worth theo Industry và sắp xếp
df_industry= df[["Industry", "Net Worth"]].groupby("Industry").sum()
df_industry = df_industry.sort_values(by='Net Worth', ascending=False)
df_industry
```

**Biểu đồ cột biểu diễn top 10 lĩnh vực có tổng tài sản cao nhất**
- Mỗi cột đại diện cho 1 lĩnh vực
- Tổng tài sản của lĩnh vực được biểu diễn bằng độ cao của cột

```python
# Tạo DataFrame gồm top 10 lĩnh vực có tổng tài sản lớn nhất
df_topindustry = df_industry[:10]
```

```python
# Vẽ biểu đồ cột
plt.bar(x=df_topindustry.index, height=df_topindustry['Net Worth'])
# Xoay label ở trục x 90 độ
plt.xticks(rotation=90)

plt.ylabel('Net Worth ($B)')
plt.title('Top 10 Net Worth by Industry')
plt.show()
```

# **Biểu đồ tần suất - Histogram**
- Sử dụng nhiều cột (bar) để biểu diễn phân phối của dữ liệu
- Chiều cao của cột thể hiện số lần xuất hiện của giá trị cột đó

```python
# Biểu đồ tần suất biểu diễn phân phối tuổi của các tỉ phú
plt.hist(
    x=df["Age"],
    edgecolor='white'
    )
plt.title("Forbes's 2024 Billionaires Age Distribution")
plt.ylabel("Count")
plt.xlabel("Age")
plt.show()
```

# **Biểu đồ phân tán - Scatter Plot**
- Là dạng biểu đồ phân phối
- Sử dụng các điểm dữ liệu (point) để biểu diễn mối quan hệ mạnh - yếu giữa các biến

Khi xem xem xét biểu đồ phân tán, cần xét:
- Các điểm dữ liệu tụ lại hay tán ra
- Phân phối dương hay âm (sang trái hay sang phải)

**Biểu đồ phân tán biểu diễn mỗi quan hệ giữa tuổi và tổng tài sản**
- Mỗi điểm dữ liệu sẽ thể hiện cả 2 giá trị: tuổi, tổng tài sản
- Tập hợp các điểm dữ liệu biểu diễn được phân phối giữa tuổi tác và tổng tài sản

```python
plt.scatter(
    x=df["Age"],
    y=df["Net Worth"],
    s=30,
    linewidths=0.1,
    edgecolors='black'
)
plt.title("2024 Billionaires's Age w.r.t Net Worth")
plt.xlabel("Age")
plt.ylabel("Net Worth ($B)")
plt.show()
```

# **Biểu đồ hộp - Box Plot**
- Tác dụng: so sánh dữ liệu với giá trị trung bình

```python
import pycountry_convert as pc
```

```python
# Lấy tên châu lục thông qua tên quốc gia
def get_continent(country_name):
    try:
        country_code = pc.country_name_to_country_alpha2(country_name)
        continent_code = pc.country_alpha2_to_continent_code(country_code)
        continent_name = pc.convert_continent_code_to_continent_name(continent_code)
        return continent_name

    except:
        print(country_name)
        return "Unknown"

df["Country | Territory"] = df["Country | Territory"].replace({"Eswatini (Swaziland)":"Swaziland"})
df["Continent"] = df["Country | Territory"].apply(get_continent)
```

**Biểu đồ hộp biểu diễn phân phối top tài sản (>50 tỷ) theo châu lục**
- Mỗi hộp thể hiện phân phối của 1 châu lục
- Đường màu cam ở giữa thể hiện giá trị trung bình (median)

```python
grouped = df[df["Net Worth"]>50].groupby('Continent')['Net Worth'].apply(list)
plt.boxplot(grouped, tick_labels=grouped.index)
plt.xlabel('Continent')
plt.ylabel('Net Worth')
plt.title("Distribution of top richest billionaires (Net Worth >$50B ) grouped by continent")
plt.show()
```
